**Note 1**: in order to have a similar shape, we need to select the amplitudes and phases carefully. One way to do this is to define the harmonic signal as follows:

$$y(t) = \sum_{k=1}^K A_k\cos\left(2\pi kf_0 t + k \phi - (k-1)\pi/2 \right), $$

where $K$ is the number of harmonics, $f_0$ is the fundamental frequency, $A_k$ is the amplitude (weight) of each harmonic and $\phi$ is the phase of the signal (defined in Lab1 Ex 2.4).

**Note 2**:
Define the $A_k$ values relative to the fundamental frequency. This means to define $A_1=1$ and the others less than 1. You can use Audacity to plot the spectrum of the fragment selected of the reference audio and measure the relative amplitudes of the harmonics.

**Note 3**: Normalize the amplitude of the signal by the same amplitude of the reference. For instance, if the amplitude of the reference signal is 0.33, you can normalize the syntesized signal by first dividing by its maximum and then multiplyng by 0.33:


````python
import numpy as np

filepath = "./sis1_group102/592442__baconation__guitar-pluck.mp3"
ref, fs = load_audio(filepath)

# Time window
t_start = 0.044
t_end = 0.06
start_idx = int(t_start * fs)
end_idx = int(t_end * fs)

ref_seg = ref[start_idx:end_idx]
t = np.arange(len(ref_seg)) / fs + t_start

# --- Harmonic synthesis parameters ---
f0 = 500.0
stretch_factor = 2.086     # sets period ≈ 0.004171 s
f0_stretched = f0 / stretch_factor

K = 6
A = np.array([1.0, 0.55, 0.28, 0.15, 0.08, 0.05])   # smoother crest
phi = np.array([np.pi/2, np.pi/2.5, np.pi/4.5, np.pi/4.5, np.pi/2, np.pi/4])

phi_shift = -np.pi / 10    # keep same alignment as your best version

k = np.arange(1, K + 1)[:, np.newaxis]

y = np.sum(
    A[:, np.newaxis] * np.cos(2 * np.pi * k * f0_stretched * t + phi[:, np.newaxis] + phi_shift),
    axis=0
)

# normalize amplitude to around 0.04
target_amp = 0.04
y = target_amp * y / np.max(np.abs(y))

# embed back into reference-length vector
x = np.zeros_like(ref)
x[start_idx:end_idx] = y

plot_signals([ref, x], fs, t_start=t_start, t_end=t_end)
````

The synthesized signal sounds purer, static and more regular, while the reference, the guitar pluck, is richer, realistic and more dynamic. Basically, The guitar body adds color and depth, while the synthesized signal is dry.
